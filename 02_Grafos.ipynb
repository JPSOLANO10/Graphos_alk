```python
import graphviz
import pandas as pd
from tqdm import tqdm
import matplotlib.pyplot as plt
import numpy as np
import matplotlib.colors as mcolors
from ipywidgets import interact, Dropdown, Layout
```

```python
datos = pd.read_excel(r"D:\OneDrive - Emtelco\Desarollo\Alkosto\codigo\base_menu.xlsx")
```

```python
datos.head()
```

```python
resultado = datos.groupby(["op_text","op_text_final"]).count()[["idtransaccion"]].sort_values("idtransaccion",ascending=False).reset_index()
# Definir el color en RGB
```

```python
rojo = 191
verde = 127
azul = 146

# Convertir a formato hexadecimal
color_hex = '#{:02x}{:02x}{:02x}'.format(rojo, verde, azul)
cmap = plt.cm.get_cmap('YlOrRd')  # página de colorcitos: https://matplotlib.org/stable/users/explain/colors/colormaps.html
norm = mcolors.Normalize(vmin=np.log(resultado['idtransaccion'].min()), vmax=np.log(resultado['idtransaccion'].max()))
u = graphviz.Digraph('unix', filename='base_grafo_menu.gv', node_attr={'color': color_hex, 'style': 'filled'})
u.attr(size='500,500')

u.attr(bgcolor='#{:02x}{:02x}{:02x}'.format(250, 247, 247))
for i in tqdm(range(len(resultado))):
    u.edge(resultado["op_text"][i],resultado["op_text_final"][i],label=str(resultado["idtransaccion"][i]))
    frequency = resultado["idtransaccion"][i]

for op_text_final_value in resultado['op_text_final'].unique():
    frequency = resultado[resultado['op_text_final'] == op_text_final_value]['idtransaccion'].sum()
    color = mcolors.to_hex(cmap(norm(np.log(frequency))))
    u.node(op_text_final_value, fillcolor=color+"80")


u.view()
```
